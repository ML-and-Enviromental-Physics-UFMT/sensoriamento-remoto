## 1. Intuição

Comparar datas exige **co-registro**. Métodos:

1. Diferença absoluta + limiar.
2. PCA bitemporal.
3. MAD / iMAD (diferenças canônicas iteradas).

Abaixo: simulação com **uma** imagem raster sintético (mesmo fluxo de `ex09_2.py`).

```python
from pathlib import Path
import numpy as np
import matplotlib.pyplot as plt
import rasterio

REPO = next(p for p in (Path.cwd(), *Path.cwd().parents) if (p / "study_utils").is_dir())
EXEMPLO = REPO / 'data/exemplos/exemplo_sintetico.tif'
OUT = REPO / "conteudo/series-temporais/figuras"; OUT.mkdir(exist_ok=True)

with rasterio.open(EXEMPLO) as ds:
    before = ds.read(4).astype(np.float64)
after = before.copy()
r, c = before.shape
after[r//3:2*r//3, c//3:2*c//3] *= 0.55
diff = np.abs(after - before)
thr = np.percentile(diff, 95)
change = diff >= thr

fig, ax = plt.subplots(1, 3, figsize=(12, 4))
ax[0].imshow(before, cmap='gray'); ax[0].set_title('Antes')
ax[1].imshow(after, cmap='gray'); ax[1].set_title('Depois (simulado)')
ax[2].imshow(change, cmap='gray'); ax[2].set_title('Mudança')
for a in ax: a.axis('off')
fig.savefig(OUT/'nb09_sim_change.png', dpi=120, bbox_inches='tight')
plt.show()
print('limiar', thr)
```
