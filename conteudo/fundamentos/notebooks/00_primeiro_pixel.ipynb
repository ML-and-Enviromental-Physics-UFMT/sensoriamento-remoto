# Seu primeiro pixel

Vamos representar duas bandas de uma cena inventada com duas linhas e duas colunas. Os valores são reflectâncias hipotéticas, sem unidade. Não há imagem de satélite real, coordenadas ou validação de cobertura do solo neste exercício.

A prática usa só Python. Execute na ordem. Se nunca programou, comece pelas células abaixo e altere um número por vez.


## 1. Guardar valores e acessar uma posição

`=` guarda um valor em um nome. Colchetes criam listas. Uma lista de listas representa aqui as linhas de uma banda. Python conta posições a partir de zero. Por isso, `[0][1]` indica primeira linha e segunda coluna. `print` mostra um resultado.

Antes de executar, procure na lista o valor que será exibido.


```python
vermelho = [[0.10, 0.20], [0.30, 0.00]]
nir = [[0.50, 0.25], [0.35, 0.00]]

print("Vermelho na primeira linha, segunda coluna:", vermelho[0][1])
print("NIR na mesma posição:", nir[0][1])
```

Os valores esperados são 0.20 e 0.25. O vetor desse pixel contém as duas medições nessa ordem, vermelho e NIR. A posição na lista não identifica, por si só, uma coordenada geográfica.

## 2. Uma média feita à mão

`+` soma e `/` divide. Os parênteses agrupam a soma antes da divisão. A barra invertida não é necessária quando a expressão está entre parênteses.

Nesta cena, o zero é uma medição inventada válida. Em um produto real, seria necessário consultar a máscara e a definição de NoData antes de incluí-lo na média.


```python
media_vermelho = (
    vermelho[0][0] + vermelho[0][1] + vermelho[1][0] + vermelho[1][1]
) / 4
print("Média do vermelho:", media_vermelho)
```

O resultado é aproximadamente 0.15. Pequenas diferenças nas últimas casas podem aparecer pela representação dos números no computador.

## 3. Calcular NDVI e tratar uma conta indefinida

Uma função, criada com `def`, reúne instruções para repetir a mesma conta com entradas diferentes. `return` devolve o resultado. O bloco `if` trata aqui o caso em que não podemos dividir. A indentação faz parte da sintaxe do Python.

`None` representa ausência de resultado nesta função. Não é zero. Um NDVI igual a zero seria um resultado numérico válido quando as duas bandas fossem iguais e a soma fosse diferente de zero.


```python
def calcular_ndvi(vermelho_pixel, nir_pixel):
    soma = nir_pixel + vermelho_pixel
    if soma == 0:
        return None
    return (nir_pixel - vermelho_pixel) / soma

print("Primeiro pixel:", calcular_ndvi(vermelho[0][0], nir[0][0]))
print("Segundo pixel:", calcular_ndvi(vermelho[0][1], nir[0][1]))
print("Pixel com soma zero:", calcular_ndvi(vermelho[1][1], nir[1][1]))
```

Você deve obter aproximadamente 0.67, 0.11 e `None`. Essa função trata somente o denominador zero. Uma aplicação real também precisa aplicar máscaras de qualidade e as conversões documentadas do produto.

## 4. Exibir não é calibrar

Vamos transformar a banda vermelha para que seu menor valor vire 0 e seu maior vire 1. `min` e `max` encontram os extremos dos quatro valores listados. A operação serve para estudar contraste, não para produzir reflectância calibrada.

Preveja o que acontece com o valor original 0.30.


```python
valores = [vermelho[0][0], vermelho[0][1], vermelho[1][0], vermelho[1][1]]
menor = min(valores)
maior = max(valores)
valor_exibido = (vermelho[1][0] - menor) / (maior - menor)
print("Medição original:", vermelho[1][0])
print("Valor para exibição:", valor_exibido)
```

O original continua 0.30 e o valor para exibição é 1.0. A expressão calcula outro número; ela não modifica a lista `vermelho`. Em uma banda constante, esse stretch teria denominador zero e precisaria de tratamento próprio.

## 5. Sua vez

Antes de rodar a célula seguinte, calcule o NDVI para vermelho 0.20 e NIR 0.60. Depois troque apenas o NIR por 0.30 e preveja o sentido da mudança. Você não precisa alterar as listas anteriores.


```python
vermelho_teste = 0.20
nir_teste = 0.60
print("NDVI do teste:", calcular_ndvi(vermelho_teste, nir_teste))
```
