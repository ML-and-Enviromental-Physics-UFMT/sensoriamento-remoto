## 1. Intuição

Sem rótulos, agrupamos pixels espectralmente semelhantes (**clusters**). Útil para explorar a cena, gerar candidatos a classes e pré-segmentar. A legenda (nome da cobertura) vem depois, por interpretação.

- **k-means**: centróides + distância euclidiana.
- **EM/GMM**: mistura de gaussianas (atribuição soft).
- **SOM**: mapa topológico 2D.

```python
from pathlib import Path
import numpy as np
import matplotlib.pyplot as plt
import rasterio
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler

REPO = next(p for p in (Path.cwd(), *Path.cwd().parents) if (p / "study_utils").is_dir())
EXEMPLO = REPO / 'data/exemplos/exemplo_sintetico.tif'
OUT = REPO / "conteudo/classificacao/agrupamento/figuras"; OUT.mkdir(exist_ok=True)

with rasterio.open(EXEMPLO) as ds:
    img = ds.read().astype(np.float64)
X = img.reshape(img.shape[0], -1).T
sc = StandardScaler().fit(X)
Xs = sc.transform(X)
idx = np.random.default_rng(0).choice(len(Xs), min(25000, len(Xs)), replace=False)
km = KMeans(n_clusters=5, n_init=10, random_state=0).fit(Xs[idx])
labels = km.predict(Xs).reshape(img.shape[1], img.shape[2])

plt.figure(figsize=(5,5))
plt.imshow(labels, cmap='tab10')
plt.title('k-means K=5'); plt.axis('off')
plt.savefig(OUT/'nb08_kmeans.png', dpi=120, bbox_inches='tight')
plt.show()
```
